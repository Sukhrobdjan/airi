```python
import numpy as np 
arr = np.array([1, 2, np.nan, 4, 5, np.nan])
np.isnan(arr)
print(np.isnan(arr)) # True where NaN, False otherwise
print(~np.isnan(arr)) # True where not NaN
print(arr[~np.isnan(arr)]) # array withou t NaN values
print(arr[np.isnan(arr)]) # array with only NaN values
```
